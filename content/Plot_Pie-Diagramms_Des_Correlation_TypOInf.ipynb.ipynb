```python
# ==============================
# PART 2 — PLOTTING
# Plot 5x3 pies for 15 most relevant Description clusters vs Typ OInf
# ==============================

import pandas as pd
import matplotlib.pyplot as plt

# ------------------------------
# CONFIG
# ------------------------------
sep = ";"
corr_file = "description_typOInf_correlations.csv"
max_clusters = 15
max_types_per_cluster = 10

# ------------------------------
# LOAD DATA
# ------------------------------
corr_df = pd.read_csv(corr_file, sep=sep)
if corr_df.empty:
    raise ValueError("Correlation file is empty — run Part 1 first.")

# Get top N clusters by Description_Count
top_clusters = corr_df.groupby("Description_Cluster")["Description_Count"].max().nlargest(max_clusters).index
plot_df = corr_df[corr_df["Description_Cluster"].isin(top_clusters)]

# ------------------------------
# CREATE PIE PLOTS
# ------------------------------
fig, axes = plt.subplots(5, 3, figsize=(18, 20))
axes = axes.flatten()

for i, (cluster, subdf) in enumerate(plot_df.groupby("Description_Cluster")):
    if i >= len(axes): break
    subdf_sorted = subdf.sort_values("Typ_Count", ascending=False).head(max_types_per_cluster)
    axes[i].pie(subdf_sorted["Typ_Count"], labels=subdf_sorted["Typ_OInf"], autopct="%1.1f%%", startangle=90)
    axes[i].set_title(f"Cluster {i+1}: {cluster[:60]}...", fontsize=10)

# Hide unused axes
for j in range(i+1, len(axes)):
    axes[j].axis("off")

plt.suptitle("Top 15 Description Clusters — Correlation with Typ OInf", fontsize=14, y=0.92)
plt.tight_layout()
plt.savefig("description_typOInf_pies.png", dpi=300)
plt.show()
print("✅ Saved figure: description_typOInf_pies.png")
```
